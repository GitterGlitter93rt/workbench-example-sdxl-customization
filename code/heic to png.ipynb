```python
from PIL import Image, ImageOps
import pillow_heif
import os

# Enable HEIC support
pillow_heif.register_heif_opener()

input_folder = "../datasets/muren"
output_folder = "../datasets/muren_png"

os.makedirs(output_folder, exist_ok=True)

for filename in os.listdir(input_folder):
    if filename.lower().endswith(".heic"):
        input_path = os.path.join(input_folder, filename)
        output_filename = filename.rsplit(".", 1)[0] + ".png"
        output_path = os.path.join(output_folder, output_filename)

        try:
            image = Image.open(input_path)

            # Fix orientation (critical for iPhone HEIC)
            image = ImageOps.exif_transpose(image)

            # Convert to RGB (PNG supports it fine)
            image = image.convert("RGB")

            # Save as PNG (lossless)
            image.save(output_path, "PNG", compress_level=0)

            print(f"✅ Converted: {filename} → {output_filename}")

        except Exception as e:
            print(f"❌ Failed: {filename} → {e}")

print("\n🎉 Done converting HEIC → PNG")
```
